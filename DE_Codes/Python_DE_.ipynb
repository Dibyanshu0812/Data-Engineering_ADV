```python
#check prime numbers
numbers = [2,3,4,5,6,7,8,11]
def is_prime(n):
    if n < 2:
        return False
    for i in range(2,int(n**0.5)+1):
        if n%i==0:
            return False
    return True

count = sum(is_prime(x) for x in numbers)
print(count)
```
