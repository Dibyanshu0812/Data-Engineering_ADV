```python
data = [
(1,"Alice",25,"alice@email.com"),
(2,"Bob",None,"bob@email.com"),
(3,None,30,None),
(4,"David",None,None)
]

cols = ["customer_id","name","age","email"]

df = spark.createDataFrame(data,cols)

df.show()
```

### Dropping Nulls [If there is one Null, Entire column will be dropped]

```python
#drop_nulls = df.dropna()
drop_nulls = df.na.drop()
drop_nulls.display()
```

### Dropping Nulls [If there is Null in one column, then delete]

```python
df_drop_nulls = df.na.drop(subset=["age"])
df_drop_nulls.display()
```

### Fill mising value Fill na 

```python
df_fill_data = df.fillna({"age":"-1"}).fillna({"email":"Not_provided"})
df_fill_data.display()
```

### Gold Logic after handling nulls

```python
from pyspark.sql.functions import col, when
df_gold_data = df_fill_data.withColumn(
  "Customer_type", when(col("email")=="Not_provided", "Guest").otherwise("Registered")
)
df_gold_data.display()
```

### Silver Handling Null

```python
from pyspark.sql.functions import coalesce, lit
df_handling_nulls_colease = df.withColumn("email", coalesce("email", lit("Not_provided")))

df_handling_nulls_colease.display()
```
